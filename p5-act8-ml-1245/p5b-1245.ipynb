```python
# Gasto diario en un diccionario key ,dia gasto
# values lista de dias listas gastos
datos = {
    "Dia": ["Lunes", "Martes", "Miercoles", "Jueves", "Viernes"],
    "Gasto": [50, 35, 80, 45, 60]
}

# DataFrame tabla o marco de datos
df = pd.DataFrame(datos)
print("=== GASTOS SEMANALES ===")
print(df)
```

```python
print("=== El total de gastos y el promerio de gastos ===")
print("Gasto total:", df["Gasto"].sum())
print("Gasto promedio:", df["Gasto"].mean())
print("\nDías con gasto mayor a $50:")
print(" Key Gastos values [50, 35, 80, 45, 60]")
print(df[df["Gasto"] > 50])

print("+-+-+-+-+-+ Jupyter Dylan NC 1245 +-+-+-+-+-+-")
```
